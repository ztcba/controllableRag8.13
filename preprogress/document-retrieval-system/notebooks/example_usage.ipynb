# Document Retrieval System Example Usage

This notebook demonstrates how to use the Document Retrieval System to ingest Markdown documents, create a FAISS vector store, and retrieve documents using a hybrid retriever.

```python
# Import necessary modules
from rag_pipeline.ingest import ingest_documents
from rag_pipeline.components.retriever_factory import create_retriever

# Define paths
DATA_PATH = 'data/raw'
VECTOR_STORE_PATH = 'data/processed/vector_store.faiss'

# Ingest documents
ingest_documents(DATA_PATH, VECTOR_STORE_PATH)

# Create a retriever
retriever = create_retriever(
    vector_store_path=VECTOR_STORE_PATH,
    embedding_model='text-embedding-model',  # Specify your embedding model
    use_reranking=True,  # Set to True to enable re-ranking
    top_k=5  # Number of documents to return
)

# Example query
query = 'What is the impact of climate change on agriculture?'
results = retriever.retrieve(query)

# Display results
for result in results:
    print(result['document'], '\nScore:', result['score'])
```
